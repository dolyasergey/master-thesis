```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from helper import load_params_csv
from calibrate import calibrate_eta
```

```python
calibrate_params = load_params_csv('params/IP_calibrate.csv')

sigma = float(calibrate_params["sigma"]) 
tau = float(calibrate_params["tau"])
rho = float(calibrate_params["r_bar"]) + float(calibrate_params["pi"])
mu_tv = 0.02
dv = float(calibrate_params["debt"])

T = 13
n_steps = T * 4
n_paths = 1000

sigma_market = 0.324

IP_eta_star, info = calibrate_eta(
    sigma_market = sigma_market,
    sigma        = sigma,
    params       = calibrate_params,
    n_steps      = n_steps,
    n_paths      = n_paths,
    T            = T,
    rho          = rho,
    mu_tv        = mu_tv,
    tau          = tau,
    dv_ratio     = dv,
    seed         = 99,
    verbose      = False,
)

sigma_error = info['sigma_E_at_eta_star'] - info['sigma_market']

print(f"Calibrated eta for IP = {IP_eta_star:.4f}")
print(f"Estimation Error = {sigma_error:.4f}")
```
